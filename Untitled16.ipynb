```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

# Generate a synthetic 2D dataset with 3 clusters
X, y = make_blobs(n_samples=300, centers=3, n_features=2, random_state=42, cluster_std=0.60)

# Visualize our input dataset
plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], X[:, 1], c='blue', alpha=0.6, edgecolors='k')
plt.title("Input Dataset (Unsupervised Learning)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True)
plt.show()
```

```python
# SOM Parameters
epochs = 100
learning_rate_initial = 0.5
num_neurons = 10

# Initialize weights randomly within the range of our dataset
np.random.seed(42)
input_dim = X.shape[1]
weights = np.random.uniform(np.min(X), np.max(X), (num_neurons, input_dim))

# Function to find the Best Matching Unit (BMU)
def find_bmu(sample, weights):
    distances = np.sum((weights - sample) ** 2, axis=1)
    return np.argmin(distances)

# Training the SOM
for epoch in range(epochs):
    # Decay learning rate linearly
    learning_rate = learning_rate_initial * (1 - epoch / epochs)

    for sample in X:
        # 1. Find the BMU (winning neuron)
        bmu_idx = find_bmu(sample, weights)

        # 2. Update weights of BMU and its immediate neighbors (1D neighborhood)
        for i in range(num_neurons):
            distance_to_bmu = abs(i - bmu_idx)
            if distance_to_bmu <= 1:  # Neighborhood of size 1
                # Influence factor based on topological distance
                influence = np.exp(-distance_to_bmu)
                weights[i] += learning_rate * influence * (sample - weights[i])

print("SOM Training Completed!")
```

```python
plt.figure(figsize=(10, 6))
# Plot original data points
plt.scatter(X[:, 0], X[:, 1], c='lightgray', alpha=0.8, edgecolors='none', label='Data Points')

# Plot the SOM neurons and the grid line connecting them
plt.plot(weights[:, 0], weights[:, 1], 'ro-', label='SOM Neurons / Topology', linewidth=2, markersize=8)

plt.title("Trained Self-Organizing Map (SOM)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True)
plt.show()
```
