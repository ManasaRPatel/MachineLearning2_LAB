```python
import pandas as pd

data = {
    'CGPA': ['High', 'High', 'Medium', 'Low', 'High', 'Medium'],
    'Attendance': ['High', 'High', 'High', 'Low', 'High', 'High'],
    'Backlogs': ['No', 'No', 'No', 'Yes', 'Yes', 'No'],
    'Income': ['Low', 'Medium', 'Low', 'High', 'Low', 'Medium'],
    'Eligible': ['Yes', 'Yes', 'Yes', 'No', 'No', 'Yes']
}

df = pd.DataFrame(data)

df.to_csv('scholarship.csv', index=False)

print("Dataset created successfully!")
display(df)
```

```python
import pandas as pd

df = pd.read_csv('scholarship.csv')

print("Dataset:")
display(df)

print("\nDataset shape:", df.shape)
```

```python
# Separate attributes and target
attributes = list(df.columns[:-1])
target = df.columns[-1]

# Initialize the most specific hypothesis
hypothesis = ['Ø'] * len(attributes)

print("Initial Hypothesis:", hypothesis)

# Process each training example
for _, row in df.iterrows():

    # Consider only positive examples
    if row[target] == 'Yes':

        for i, attribute in enumerate(attributes):

            if hypothesis[i] == 'Ø':
                hypothesis[i] = row[attribute]

            elif hypothesis[i] != row[attribute]:
                hypothesis[i] = '?'

        print("Positive Example:", list(row[attributes]))
        print("Updated Hypothesis:", hypothesis)
        print()

print("Final Hypothesis:", hypothesis)
```

```python
with open('find_s_result.txt', 'w') as file:
    file.write("Dataset: Student Scholarship Eligibility\n")
    file.write("Algorithm: FIND-S\n")
    file.write(f"Attributes: {attributes}\n")
    file.write(f"Final Hypothesis: {hypothesis}\n")

print("Result saved successfully!")
```
